# Model 1: one-region CGE model

```python
import numpy as np
from scipy.optimize import root
```

## Functions

### Production

```python
def output_k(A, L):
    return A * L

def unit_labour_requirement(A):
    return 1 / A
```

### Prices

```python
def price_k(w, A):
    return w / A
```

### Household

```python
def utility(Z, alpha):
    return np.prod(Z**alpha)

def income(w, L_bar):
    return w * L_bar

def budget_constraint(Z, p):
    return np.sum(p * Z)

def demand_k(alpha, I, p):
    return (alpha * I) / p
```

### Market clearing / residuals

```python
def goods_market_clearing(Y, Z):
    return Y - Z

def labour_market_clearing(L, L_bar):
    return L_bar - np.sum(L)
```

## Modelling values

```python
L_bar = 100
A = np.array([1, 2])
alpha = np.array([0.5, 0.5])

w = 1               # From equation (12)
```

## Benchmark

```python
def system_residuals(L):
    # 1. Supply from production
    Y = output_k(A, L)

    # 2. Prices & Income
    p = price_k(w, A)
    I = income(w, L_bar)

    # 3. Household demand
    Z = demand_k(alpha, I, p)

    return goods_market_clearing(Y, Z)
```

### Goods market clearing

```python
sol = root(system_residuals, [40, 60])  # Initial guess for L_1 and L_2
print("Solution for L_1 and L_2:", sol.x)
```

```python
res = system_residuals(sol.x)
print("Residuals at solution:", res)
```

### Labour market clearing

```python
labour_market_clearing(sol.x, L_bar)
```

## 4. Explain why not all market-clearing equations are independent.

We have that $I = w \bar{L}$, $p_k = \frac{w}{A_k}$ and $Y_k = A_kL_k$, then the monetary value of production in each sector equals its total wage bill $p_k Y_k = w L_k$. Substituting goods market clearing $Y_k = Z_k$ into the household budget constraint $p_1 Z_1 + p_2 Z_2 = I$ gives $w L_1 + w L_2 = w \bar{L}$. Dividing by $w$ recovers the labour market clearing condition $L_1 + L_2 = \bar{L}$, making it mathematically redundant once the other markets and budget constraints hold. Therefore, the equations are not independent.

## 5. Explain what would happen if all nominal prices and the wage were multiplied by the same positive constant.

As $p_k = \frac{w}{A_k}$, we have that nominal prices scale proportionally with wages. Income also scales with wages, and as demand is $\frac{\alpha_k I}{p_k}$, the demand stays the same when both nominal prices and the wage are multiplied by the same positive constant.
